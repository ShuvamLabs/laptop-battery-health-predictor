## Training and testing different algorithms and finding the best one

1. Import libraries

```python
import joblib
import numpy as np
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import Pipeline
from sklearn.linear_model import LinearRegression
from sklearn.tree import DecisionTreeRegressor
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import r2_score, mean_absolute_error, mean_squared_error
```

### Model 1 : All legitimate data

2. Load the Processed Data

```python
X_train_1 = joblib.load(r"C:\Users\HP\Documents\PycharmProjects\laptop-battery-health-predictor\data\processed-data\X_train.pkl")
X_test_1 = joblib.load(r"C:\Users\HP\Documents\PycharmProjects\laptop-battery-health-predictor\data\processed-data\X_test.pkl")
y_train = joblib.load(r"C:\Users\HP\Documents\PycharmProjects\laptop-battery-health-predictor\data\processed-data\y_train.pkl")
y_test = joblib.load(r"C:\Users\HP\Documents\PycharmProjects\laptop-battery-health-predictor\data\processed-data\y_test.pkl")
```

3. Make Pipelines

```python
lr_pipeline = Pipeline([
    ('scaler', StandardScaler()),
    ('regressor', LinearRegression())
])
rf_pipeline = Pipeline([
    ('scaler', StandardScaler()),
    ('regressor', RandomForestRegressor())
])
dt = DecisionTreeRegressor()
```

4. Train and Test data

4.1. Linear Regression

```python
lr_pipeline.fit(X_train_1, y_train)
predictions = lr_pipeline.predict(X_test_1)
print("MAE:", mean_absolute_error(y_test, predictions))
print("RMSE:", np.sqrt(mean_squared_error(y_test, predictions)))
print("R² Score:", r2_score(y_test, predictions))
```

4.2. Random Forest Regressor

```python
rf_pipeline.fit(X_train_1, y_train)
predictions = rf_pipeline.predict(X_test_1)
print("MAE:", mean_absolute_error(y_test, predictions))
print("RMSE:", np.sqrt(mean_squared_error(y_test, predictions)))
print("R² Score:", r2_score(y_test, predictions))
```

4.3. Decision Tree Regressor

```python
dt.fit(X_train_1, y_train)
predictions = dt.predict(X_test_1)
print("MAE:", mean_absolute_error(y_test, predictions))
print("RMSE:", np.sqrt(mean_squared_error(y_test, predictions)))
print("R² Score:", r2_score(y_test, predictions))
```
